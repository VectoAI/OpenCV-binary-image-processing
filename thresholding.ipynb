```python
import numpy as np 
import cv2 as cv 
import matplotlib.pyplot as plt  
import os 
```

```python
# dynamic path 
current_path = os.getcwd()
project_dir = os.path.dirname(current_path)
image_dir = os.path.join(project_dir, 'images')
```

simple thresholding by get binary image

```python
image1_path = os.path.join(image_dir, 'Lusin-1.jpg')
image = cv.imread(image1_path, 0)
new_image = image.copy()
new_image[new_image >= 127] = 255
new_image[new_image < 127] = 0

plt.figure(figsize=(10, 10))

plt.subplot(121)
plt.imshow(image, cmap='gray')
plt.subplot(122)
plt.imshow(new_image, cmap='gray')
```

5 methods for get binary image 

```python
image2_path = os.path.join(image_dir, 'plate.jpg')
image = cv.imread(image2_path, 0)

ret, new_image_th1 = cv.threshold(image, 127, 255, cv.THRESH_BINARY)
ret, new_image_th2 = cv.threshold(image, 127, 255, cv.THRESH_BINARY_INV)
ret, new_image_th3 = cv.threshold(image, 127, 255, cv.THRESH_TOZERO)
ret, new_image_th4 = cv.threshold(image, 127, 255, cv.THRESH_TOZERO_INV)
ret, new_image_th5 = cv.threshold(image, 127, 255, cv.THRESH_TRUNC)

plt.figure(figsize=(10, 5))

plt.subplot(231)
plt.imshow(image, cmap='gray')
plt.subplot(232)
plt.imshow(new_image_th1, cmap='gray')
plt.subplot(233)
plt.imshow(new_image_th2, cmap='gray')
plt.subplot(234)
plt.imshow(new_image_th3, cmap='gray')
plt.subplot(235)
plt.imshow(new_image_th4, cmap='gray')
plt.subplot(236)
plt.imshow(new_image_th5, cmap='gray')
```
